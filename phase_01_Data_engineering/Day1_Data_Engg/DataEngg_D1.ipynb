## **Mini Project:Student data explorer**
**Objective:**

Create a Student Perfomance Explorer using Python
and pandas to analyse student data and generate a detailed report

**Dataset columns**:


name ,gender, dept ,math_score ,programming_score, attendance percentage,
total_score



**Task to perform:**


**1.Display project title**

print a formatted title with separator lines

**2.Dataset overview:**
  
   Display:
-total number of students
-total number of columns
--number of unique departments

**3.Department-wise analysis:**
Display the count of students in each department


**4.Score analysis:**
find and display:
-Highest math score
-Lowest math score
-Average math score
-Average programmming score
-Average attendance percentage


**5.Gender-wise analysis:**
Calculate and display average total score grouped by gender

**6.Top performers**
Display the top 5 students based on total_score
show:
-student Name
-Dept
-Total score


**7.Attendance analysis**
Find the students whose attendance is below 75%
display:
-name
-dept
-attendance  percentage
If all students have attendance above 75%
print a suitable message

**
Concepts to use
-pandas,Data frame, groupby(),value_counts(),mean(),max(),min(),sort_values(),
iterrows()**

mini project:student data explorer
student performance using python nd pandas,to analyse student data nd generate a detailed report,
database columns
name,gender,dept

```python
import pandas as pd

# ==================================================
#         STUDENT PERFORMANCE EXPLORER
# ==================================================

print("==================================================")
print("         STUDENT PERFORMANCE EXPLORER")
print("==================================================")

# Loading Dataset
df = pd.read_csv("student_performance (1).csv")

print("Dataset loaded successfully")

print("--------------------------------------------------")

# ==================================================
#               1. DATASET OVERVIEW
# ==================================================

print("==================================================")
print("               DATASET OVERVIEW")
print("==================================================")

# Total number of students
total_students = df.shape[0]
print("Total Number of Students :", total_students)

# Total number of columns
total_columns = df.shape[1]
print("Total Number of Columns :", total_columns)

# Number of unique departments
unique_departments = df["department"].nunique()
print("Number of Unique Departments :", unique_departments)

print("--------------------------------------------------")

# ==================================================
#           2. DEPARTMENT-WISE ANALYSIS
# ==================================================

print("==================================================")
print("         DEPARTMENT-WISE ANALYSIS")
print("==================================================")

department_count = df["department"].value_counts()

print("Student Count in Each Department")
print(department_count)

print("--------------------------------------------------")

# ==================================================
#                 3. SCORE ANALYSIS
# ==================================================

print("==================================================")
print("               SCORE ANALYSIS")
print("==================================================")

# Highest math score
highest_math = df["math_score"].max()
print("Highest Math Score :", highest_math)

# Lowest math score
lowest_math = df["math_score"].min()
print("Lowest Math Score :", lowest_math)

# Average math score
average_math = df["math_score"].mean()
print("Average Math Score :", round(average_math,2))

# Average programming score
average_programming = df["programming_score"].mean()
print("Average Programming Score :", round(average_programming,2))

# Average attendance percentage
average_attendance = df["attendance_percentage"].mean()
print("Average Attendance Percentage :", round(average_attendance,2))

print("--------------------------------------------------")

# ==================================================
#             4. GENDER-WISE ANALYSIS
# ==================================================

print("==================================================")
print("            GENDER-WISE ANALYSIS")
print("==================================================")

# First, calculate total_score if it's not already present
if 'total_score' not in df.columns:
    df['total_score'] = df['math_score'] + df['science_score'] + df['english_score'] + df['programming_score']

gender_average = df.groupby("gender")["total_score"].mean()

print("Average Total Score Based on Gender")
print(round(gender_average,2))

print("--------------------------------------------------")

# ==================================================
#                 5. TOP PERFORMERS
# ==================================================

print("==================================================")
print("                TOP PERFORMERS")
print("==================================================")

top_students = df.sort_values(by="total_score", ascending=False)

top_5 = top_students.head(5)

print("Top 5 Students Based on Total Score")
print(top_5[["name","department","total_score"]])

print("--------------------------------------------------")

# ==================================================
#               6. ATTENDANCE ANALYSIS
# ==================================================

print("==================================================")
print("             ATTENDANCE ANALYSIS")
print("==================================================")

low_attendance = df[df["attendance_percentage"] < 75]

if low_attendance.empty:
    print("All students have attendance above 75%")
else:
    print("Students with Attendance Below 75%")
    print(low_attendance[["name","department","attendance_percentage"]])
```
